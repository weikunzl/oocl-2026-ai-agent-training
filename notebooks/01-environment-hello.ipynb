# 01 · 环境连通

**推荐入口：浏览器中的 Jupyter Lab（不需要 IntelliJ）。**

验证三项基础能力：

1. Python 运行环境可用（建议 3.12+）
2. LangGraph / LangChain 可导入且为 1.x
3. `LLM_*` 配置生效，能成功调用大模型

## 怎么打开本 notebook

1. 在仓库根目录执行：`uv run jupyter lab`
2. 浏览器打开终端给出的 `http://localhost:8888/lab?...`
3. 打开本文件；右上角 Kernel 选择 **OOCL 2026 AI Agent**
4. 菜单 **Run → Run All Cells**，看到 `L0 env ok` 即通关


```python
from __future__ import annotations

import os
import sys
from pathlib import Path

from dotenv import load_dotenv

ROOT = Path.cwd()
if not (ROOT / "src" / "hello_agent").exists():
    for p in [ROOT, *ROOT.parents]:
        if (p / "src" / "hello_agent").exists():
            ROOT = p
            break
os.chdir(ROOT)
sys.path.insert(0, str(ROOT / "src"))
load_dotenv(ROOT / ".env")

print("cwd =", ROOT)
print("python =", sys.version.split()[0])
assert sys.version_info >= (3, 12), (
    f"需要 Python 3.12+，当前 {sys.version.split()[0]}。"
    "请在 Jupyter 右上角切换 Kernel 为 OOCL 2026 AI Agent"
)
print("LLM_BASE_URL =", os.getenv("LLM_BASE_URL"))
print("LLM_MODEL =", os.getenv("LLM_MODEL"))
print("has_LLM_API_KEY =", bool(os.getenv("LLM_API_KEY")))
```

```python
import importlib.metadata as md

import langgraph
import langchain
import langchain_core
import langchain_openai

def ver(name: str) -> str:
    try:
        return md.version(name)
    except md.PackageNotFoundError:
        return "unknown"

print("langgraph =", ver("langgraph"), "| module", getattr(langgraph, "__version__", "?"))
print("langchain =", ver("langchain"))
print("langchain-core =", ver("langchain-core"))
print("langchain-openai =", ver("langchain-openai"))

lg = ver("langgraph")
assert lg.startswith("1."), f"需要 LangGraph 1.x，当前 {lg}"
print("LangGraph 1.x import OK")
```

```python
from langchain_core.messages import HumanMessage

from hello_agent.config import get_llm

llm = get_llm()
resp = llm.invoke([
    HumanMessage(content="用一句话打个招呼，说明你是 OOCL 环境验证探针。")
])
print(resp.content)
print("L0 env ok")
```

若失败：

1. 确认浏览器 Kernel 是 **OOCL 2026 AI Agent**（不是系统 3.11）
2. 检查仓库根目录 `.env` 的 `LLM_BASE_URL` / `LLM_MODEL` / `LLM_API_KEY`
3. 改完 `.env` 后：**Kernel → Restart Kernel**，再 **Run All Cells**
